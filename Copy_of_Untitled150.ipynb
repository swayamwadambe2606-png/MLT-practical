```python
#importing necessary libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data.columns = data.columns.str.replace(' ', '_')
data.columns.tolist()
```

```python
mv =data.isnull().sum()
```

```python
data
```

```python
mv[mv>0].sort_values(ascending=False)
print(mv)
```

```python
#filling the missing values in numerical col

num_cols = data.select_dtypes(include=np.number).columns
data[num_cols]= data[num_cols].fillna(data[num_cols].median())

#task: verify if null values from numerical col are filled
```

```python
#filling the missing values in categorical col

cat_cols = data.select_dtypes(exclude=np.number).columns
#data[cat_cols]= data[cat_cols].fillna(data[cat_cols].mode())
data[cat_cols]= data[cat_cols].apply(lambda x: x.fillna(x.mode()[0]))

#task: verify if null values from categorical col are filled
```

```python
#feature selection
data_model = data[[
'SalePrice', 'Overall_Qual', 'Gr_Liv_Area', 'Garage_Cars',
                 'Total_Bsmt_SF', 'Full_Bath', 'Year_Built', 'Lot_Area'
]]

print(data_model)
```

```python
#numpy operations

data_np = data_model.to_numpy()
print("shape", data_np.shape)
print("Mean price", np.mean(data_np[:,0]))
print("Min price", np.min(data_np[:,0]))
```

```python
#filtering or analysis

expensive_houses = data_model[data_model['SalePrice']>400000]
print(expensive_houses.shape)
```

```python
#feature engineering
data_model['Price_per_sqft'] = data_model['SalePrice']/data_model['Gr_Liv_Area']
```

```python
data_model['house_age']= 2026- data_model['Year_Built']
```

```python
#correlation analysis

Correlation = data_model.corr(numeric_only=True)
print(Correlation)
```

```python
#data visualization using line chart
data_sorted = data_model.sort_values(by='Year_Built')
plt.figure(figsize=(10, 6))
plt.plot(data_sorted['Year_Built'], data_sorted['SalePrice'])
plt.xlabel('YearBuilt')
plt.ylabel('SalePrice')
plt.title('price vs year built')
plt.show()
```

```python
import plotly.express as px

# Sort the data by Year_Built for the line chart
data_sorted = data_model.sort_values(by='Year_Built')

fig = px.line(data_sorted, x='Year_Built', y='SalePrice', title='Price vs Year Built (Plotly)')
fig.update_layout(xaxis_title='Year Built', yaxis_title='SalePrice')
fig.show()
```

```python
#bar chart
grouped = data_model.groupby('Year_Built')['SalePrice'].mean()
grouped.plot(kind='bar')
plt.xlabel('YearBuilt')
plt.ylabel('SalePrice')
plt.title('price vs year built')
plt.show()
```

```python
#histogram
plt.hist(data_model['SalePrice'], bins=20)
plt.xlabel('SalePrice distribution')
plt.ylabel('Frequency')
plt.title('Histogram of SalePrice')
plt.show()
```

```python
plt.scatter(data_model['Gr_Liv_Area'], data_model['SalePrice'])
plt.title('Scatter Plot of Gr_Liv_Area vs SalePrice')
plt.xlabel('Gr_Liv_Area')
plt.ylabel('SalePrice')
plt.show()
```

```python
#ml model
from sklearn.linear_model import LinearRegression
X  = data_model[['Gr_Liv_Area', 'Garage_Cars', 'Overall_Qual']]
y = data_model['SalePrice']

model = LinearRegression()
model.fit(X, y)
```

```python
# Re-instantiate and re-fit the model to ensure it's trained on the current X
model = LinearRegression()
# Ensure X and y are correctly defined as per the model training cell (ZejvlsU2dqxC)
X  = data_model[['Gr_Liv_Area', 'Garage_Cars', 'Overall_Qual']]
y = data_model['SalePrice']
model.fit(X, y)

# Now make the prediction with the correct number of features
pred = model.predict([[2000, 7, 2]])
print(pred)
```
